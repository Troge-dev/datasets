## 1. Subcategory Deep Dive: Inland Water Resources (23 Datasets)

### 1.1 `Wastewater generation and treatment.csv` (80 rows × 15 cols) — 🥇 **TOP RECOMMENDATION**
* **Primary Variables for Bivariate:**
  * **$X$ (Driver):** `Total wastewater generated (1,000 cubic metres per day)`
  * **$Y$ (Response):** `Wastewater teated in urban wastewater treatment plants (1,000 cubic metres per day)`
* **Empirical Sample Size:** $N = 24$ countries with complete paired reporting; 79 countries reporting at least one metric.
* **Statistical Performance:** Pearson $r = 0.906$ ($R^2 = 82.04\%$, $p = 1.15 	imes 10^{-9}$); Spearman $ho = 0.889$.
* **Fitted OLS Model:** $	ext{Treated} = 338.49 + 0.2964 	imes 	ext{Generated}$ (slope $eta_1 = 0.2964$).
* **Strengths:**
  * **Exceptional Business Narrative:** The regression slope of $0.296$ proves that for every $1,000	ext{ m}^3$ increase in municipal wastewater generated, **only $pprox 296	ext{ m}^3$ is treated in urban facilities**. This leaves an untreated or unmanaged **$pprox 70\%$ infrastructure deficit gap**!
  * **High-Impact Business Question:** Justifies private capital expenditure (CAPEX), water utility build-own-operate (BOO) concessions, and municipal green bonds.
  * **Clear Bivariate Relationship:** Strong linear co-movement ($R^2 = 82\%$), satisfying the DS311 two-variable constraint.
  * **Perfect Poster Title Match:** Fits the mandatory naming format seamlessly.
* **Weaknesses & Pitfalls:**
  * **Moderate Sample Size:** Only 24 nations provide complete paired data across both specific columns (primarily European and OECD reporting nations).
  * **Right Skewness:** Highly right-skewed distribution ($	ext{Skewness} = 3.23$) driven by heavy industrial nations (Germany, Poland, Turkey).
* **Suggested Poster Title:**  
  `Analysing the growth of Municipal Wastewater Generation and its Burden on Urban Treatment Infrastructure`
* **Suggested Business Question:**  
  *"As urbanization drives municipal wastewater generation growth, does urban treatment plant capacity expand at parity, or does an escalating 70% infrastructure deficit create a compelling business case for private utility CAPEX investment?"*

---

### 1.2 `Public Water Supply.csv` (131 rows × 11 cols) — 🥈 **STRONG ALTERNATIVE**
* **Primary Variables for Bivariate:**
  * **$X$ (Driver):** `Total population supplied by water supply industry (%)` ($N = 98$)
  * **$Y$ (Response):** `Net freshwater supplied by water supply industry (mio m3)` ($N = 111$)
* **Empirical Sample Size:** $N = 80$ paired countries.
* **Statistical Performance:** Spearman $ho = 0.179$; Pearson $r = 0.305$ (after logarithmic transformation of $Y$).
* **Strengths:**
  * Substantially larger cross-national coverage ($N = 80$ vs $24$).
  * Directly links utility infrastructure coverage (%) to volumetric production ($	ext{mio m}^3$).
  * Provides excellent material for univariate diagnostics (heavily left-skewed coverage % vs heavily right-skewed volume).
* **Weaknesses & Pitfalls:**
  * Non-linear, heteroscedastic scatter plot; raw linear correlation is weak ($r = 0.106$) without log-transforming water volume.
  * Per-capita delivery columns have high missingness ($N = 34$).
* **Suggested Poster Title:**  
  `Analysing the growth of Public Water Supply Coverage and its Relationship with Industrial Freshwater Production`

### 2.1 `Municipal waste treatment (latest year).csv` (145 rows × 15 cols) — 🥇 **BEST IN WASTE DOMAIN**
* **Primary Variables for Bivariate:**
  * **$X$ (Driver):** `Municipal waste collected (1000 tonnes)` ($N = 130$)
  * **$Y$ (Response):** `Municipal waste landfilled (%)` ($N = 94$) OR `Municipal waste recycled (%)` ($N = 48$)
* **Empirical Sample Size:** $N = 94$ paired countries for Landfill %; $N = 48$ for Recycling %.
* **Statistical Performance:** Spearman $ho = -0.214$ ($p = 0.038$); Log-Pearson $r = -0.167$ ($p = 0.107$).
* **Strengths:**
  * **Outstanding Circular Economy Relevance:** Addresses whether large municipal waste streams achieve economies of scale to divert waste away from landfills into recycling.
  * **Broad Geographic Coverage:** 94 sovereign states represented across diverse income tiers.
  * **Rich Univariate Statistics:** Features extreme right-skewed waste volumes ($	ext{Skewness} = 5.40$, max over 580M tonnes) contrasted against bounded percentage distributions ($0–100\%$).
* **Weaknesses & Pitfalls:**
  * Weak linear correlation; relationship exhibits high variance and regional clustering (European nations recycle heavily despite high volume, while developing nations landfill nearly 100%).
  * Recycling rate column has missing data for nearly 50% of nations ($N = 48$).
* **Suggested Poster Title:**  
  `Analysing the growth of Municipal Waste Generation and its Impact on Landfill Dependency: A Global Circular Economy Assessment`
* **Suggested Business Question:**  
  *"Does exponential growth in municipal waste generation create viable economies of scale for circular recycling infrastructure, or does it overwhelm municipal waste systems and reinforce carbon-intensive landfill reliance?"*

## 3. Subcategory Deep Dive: Energy & Minerals (6 Datasets)

### 3.1 `Energy Indicators.csv` (246 rows × 24 cols) — 🥇 **BEST SAMPLE SIZE IN ENTIRE REPOSITORY**
* **Primary Variables for Bivariate:**
  * **$X$ (Driver):** `Energy supply per capita (gigajoules per capita)` ($N = 217$)
  * **$Y$ (Response):** `Contribution of renewables to electricity production (%)` ($N = 215$)
* **Empirical Sample Size:** $N = 215$ paired sovereign states and territories.
* **Statistical Performance:** Spearman $ho = -0.132$ ($p = 0.054$); Pearson $r = -0.044$ ($p = 0.525$).
* **Strengths:**
  * **Massive Sample Size:** Over 215 global jurisdictions with clean, complete data (under $12\%$ missingness).
  * **Strong ESG & Corporate Decarbonization Theme:** Tests whether wealthy, energy-intensive economies lead the world in renewable electricity adoption.
  * **Surprising Empirical Discovery:** Correlation is near zero / slightly negative ($ho = -0.132$), proving that energy consumption growth is **not automatically matched by renewable scaling** (fossil fuels still dominate high-consumption economies).
* **Weaknesses & Pitfalls:**
  * Zero linear correlation means an OLS linear model will yield an $R^2 pprox 0\%$, which can be challenging to defend if a student doesn't understand that a flat line is also an important analytical discovery.
* **Suggested Poster Title:**  
  `Analysing the growth of National Energy Demand and the Scalability of Renewable Electricity Production`
* **Suggested Business Question:**  
  *"As national per capita energy demand grows, are economies successfully scaling renewable electricity production proportionally, or does fossil-fuel lock-in undermine global corporate ESG clean-energy targets?"*

---

### 3.2 `Energy intensity measured in terms of primary energy and GDP.csv` (196 rows × 20 cols)
* **Variables:** Energy Intensity across years 2000 to 2017 ($N = 196$ countries across all 18 years).
* **Strengths:** 100% complete data with zero nulls across 196 countries. Measures economic efficiency (energy consumed per unit of GDP).
* **Weaknesses:** This is an efficiency ratio over time, not two distinct physical variables. To use it for bivariate analysis, you would have to regress Year vs Energy Intensity or merge with another file.
* **Verdict:** 🥈 Excellent for time-series, moderate for 2-variable cross-sectional EDA.

* **Strengths:** 6,012 historical data points.

## 4. Subcategory Deep Dive: Air and Climate (7 Datasets)

---

### 4.1 `CO2_Emissions.csv` (191 rows × 35 cols) — 🥇 **BEST IN CLIMATE DOMAIN**
* **Primary Variables for Bivariate:**
  * **Option A (Growth vs Intensity):**
    * $X$: `% change since 1990` ($N = 89$ countries — literal growth metric!)
    * $Y$: `CO2 emissions per capita, latest year` ($N = 190$)
  * **Option B (Total Output vs Intensity):**
    * $X$: `CO2 emissions, latest year` ($N = 190$)
    * $Y$: `CO2 emissions per capita, latest year` ($N = 190$)
* **Statistical Performance:**
  * Option A (Growth vs Per Capita): Spearman $ho = -0.126$, Pearson $r = -0.099$ ($N = 89$).
  * Option B (Total vs Per Capita): Spearman $ho = 0.606$ ($p = 1.95 	imes 10^{-20}$); Log-Pearson $r = 0.539$ ($N = 190$).
* **Strengths:**
  * **Literal Growth Column Included:** The `% change since 1990` column explicitly measures emissions growth, making it a natural fit for `"Analysing the growth of..."`.
  * **High Global Repercussions:** Directly informs carbon border adjustment mechanisms (CBAM), carbon tax exposure, and corporate emissions auditing.
  * **Strong Correlation on Total vs Per Capita:** Spearman $ho = 0.606$ provides a clean, statistically significant relationship.
* **Weaknesses & Pitfalls:**
  * `% change since 1990` is missing for 101 countries ($53\%$ missingness), leaving $N = 89$.
  * Extreme outliers (China, USA, India have massive aggregate totals, while Qatar, Bahrain, Kuwait dominate per-capita intensity).
* **Suggested Poster Title:**  
  `Analysing the growth of Carbon Emissions and its Divergence between Total National Output and Per Capita Intensity`
* **Suggested Business Question:**  
  *"How does the historical growth of national CO₂ emissions decouple from per-capita carbon intensity, and what are the competitive liabilities for manufacturing supply chains facing international carbon taxation?"*

---

### 4.2 `GHG_Emissions_by_Sector.csv` (192 rows × 14 cols)
* **Variables:** Breakdown of greenhouse gases into Energy (transport), Industrial processes, Agriculture, and Waste.
* **Strengths:** 192 countries with percentage breakdowns across all 4 key economic sectors.
* **Weaknesses:** Absolute tonnage columns have over $65–85\%$ missing values; only the percentage-share columns are complete. Regressing two sectoral percentages against each other (e.g., Energy % vs Agriculture %) creates an artificial negative correlation because percentage compositions sum to 100%.
* **Verdict:** ⚠️ Risk of compositional data distortion.

### 5.2 Forests (`Forest Area.csv` — 237 rows × 11 cols)
* **Variables:** Forest Area in 1990, 2000, 2010, 2015, 2020 ($10^3	ext{ ha}$) and Land Area ($10^3	ext{ ha}$).
* **Strengths:** High completeness ($N = 236$) across 30 years of deforestation monitoring.
* **Weaknesses:** Forest area strongly scales with country geographic size, creating a trivial geometric correlation ($r pprox 0.99$ with total land area).
* **Verdict:** ⚠️ Trivial relationship unless transformed into net percentage deforestation rate.

### 5.3 Natural Disasters (4 Datasets: Climatological, Geophysical, Hydrological, Meteorological)
* **Variables:** Decadal occurrence counts (1990-1999, 2000-2009, 2010-2019), total deaths, persons affected, total damage ($US\$).
* **Strengths:** Excellent disaster frequency data across 175 countries.
* **Weaknesses:** Highly zero-inflated and discrete count data (Poisson-distributed). Monetary damage columns have high missingness ($> 60\%$). Not suited for standard continuous linear bivariate regression.
* **Verdict:** ❌ Not recommended for continuous bivariate EDA.

### 5.4 Governance & Biodiversity

* **`Marine protected areas.csv` & `Terrestrial_Marine protected areas.csv`:**
  * Only contain a single percentage column. Impossible to do bivariate analysis without merging external datasets.
